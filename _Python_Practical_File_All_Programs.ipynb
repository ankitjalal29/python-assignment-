## Question 1: ATM Withdrawal

Check the PIN, then check whether the amount is a multiple of 100 and the balance is sufficient. Withdraw the amount if all conditions are correct.

```python
pin = 1234
balance = 5000

user_pin = int(input("Enter your PIN: "))

if user_pin == pin:
    amount = int(input("Enter withdrawal amount: "))

    if amount % 100 == 0:
        if amount <= balance:
            balance = balance - amount
            print("Withdrawal Successful")
            print("Remaining Balance:", balance)
        else:
            print("Insufficient Balance")
    else:
        print("Amount must be a multiple of 100")
else:
    print("Invalid PIN")
```

```python
m1 = int(input("Enter marks of Subject 1: "))
m2 = int(input("Enter marks of Subject 2: "))
m3 = int(input("Enter marks of Subject 3: "))

if m1 >= 40 and m2 >= 40 and m3 >= 40:
    total = m1 + m2 + m3
    percentage = (total / 300) * 100

    print("Percentage:", percentage)

    if percentage >= 75:
        print("Distinction")
    elif percentage >= 60:
        print("First Division")
    elif percentage >= 50:
        print("Second Division")
    else:
        print("Pass")
else:
    print("Fail")
```

## Question 1: ATM - Advanced Challenge

Check PIN, account status, positive amount, multiple of 100, sufficient balance, minimum remaining balance of Rs. 500 and extra verification for amounts above Rs. 20,000.

```python
correct_pin = "1234"
account_active = True
balance = 30000

pin = input("Enter your PIN: ")

if pin != correct_pin:
    print("Invalid PIN")

elif account_active == False:
    print("Account is inactive")

else:
    amount = int(input("Enter withdrawal amount: "))

    if amount <= 0:
        print("Withdrawal amount must be positive")

    elif amount % 100 != 0:
        print("Amount must be a multiple of ₹100")

    elif amount > balance:
        print("Insufficient Balance")

    elif balance - amount < 500:
        print("Minimum balance of ₹500 must remain")

    elif amount > 20000:
        verification = input("Enter verification code: ")

        if verification == "123456":
            balance = balance - amount
            print("Withdrawal Successful")
            print("Remaining Balance: ₹", balance)
        else:
            print("Extra Verification Failed")

    else:
        balance = balance - amount
        print("Withdrawal Successful")
        print("Remaining Balance: ₹", balance)
```
